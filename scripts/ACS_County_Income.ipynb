## ACS County Income

The goal for this notebook is to scrape data from American Census Survey on the household median income of county in California from 2017-21 period to 2020-24. 

```python
# Import typical packages
from pathlib import Path
import pandas as pd
```

```python
# file path to import data
root_fldr = Path.cwd().parent
raw_fldr = root_fldr / 'data' / 'raw'
processed_fldr = root_fldr / 'data' / 'processed'
```

```python
#The tidycensus package; install if necessary
try:
    import pytidycensus as tc
except ImportError:
    %pip install pytidycensus 
    import pytidycensus as tc
```

```python
# Apply the API key (moving to an environment variable in Step 2)

tc.set_census_api_key("27dee4c3bbfc01f87405787b9586805edf900160")

```

```python
years = [2021, 2022, 2023, 2024]

ca_income_data = pd.DataFrame()
for year in years:
    income_data = tc.get_acs(
        geography = "county",
        variables = "B19013_001",
        state = "CA",
        year = year,
        survey = "acs5",
        geometry = False)
    income_data['year'] = year
    ca_income_data = pd.concat([ca_income_data, income_data], ignore_index=True)
```
