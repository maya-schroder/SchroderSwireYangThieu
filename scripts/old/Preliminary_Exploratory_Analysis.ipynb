### 5. FAIR Plan Policies by year

The FAIR Plan is the core of our project, so this table summarizes `policies` for each year: how many ZIP codes are covered, the total number of policies, and the mean and median per ZIP code. Comparing the mean with the median shows whether a few ZIP codes with very large counts pull the average up.

```python
# Summarize several variables by year
year_summary = (
    cfp_income
    .groupby("Year")
    .agg(
        zip_code_records=("zip_code", "count"),
        policies_sum=("policies", "sum"),
        policies_mean=("policies", "mean"),
        policies_median=("policies", "median")
    )
    .reset_index()
)

year_summary.head()
```

Based on the table, from 2021-24, the coverage is mostly consistent between 1548 and 1570. The policies roughly double from 238393 in 2021 to 2024 with 454259. Interestingly, comparing the mean and median, policies per ZIP are strongly right-skewed, though less so by 2024. 

### 6. Limitations

From the intial inspection, we can see there are two limitations that stood out. The first is that there are zip code that has 0 policies and the second is there is a high margin of errors for income in some zip codes. So by doing this, we inspect how many of these are there. 

```python
# Select records that have zero policies
zero_policies = cfp_income.loc[cfp_income["policies"] == 0]
zero_policies.shape
```

```python
# Select records that have high margin of error (MOE)
high_MOE = cfp_income.loc[cfp_income["Income_MOE"] > cfp_income["Median_income"] / 2]
high_MOE.shape
```
